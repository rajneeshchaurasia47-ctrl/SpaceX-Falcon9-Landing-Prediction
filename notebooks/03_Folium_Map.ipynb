# Folium map with three launch sites
Data: IBM Skills Network `dataset_part_2.csv` (90 rows). Remote API/Wikipedia collection is not claimed successful.


```python
import pandas as pd,folium,math
x=pd.read_csv('dataset_part_2.csv')
g=x.groupby('LaunchSite').agg(Latitude=('Latitude','mean'),Longitude=('Longitude','mean'),launches=('Class','size'),successes=('Class','sum')).reset_index()
m=folium.Map(location=[31.7,-101],zoom_start=4,tiles=None,control_scale=True)
for _,r in g.iterrows():
 pct=100*r.successes/r.launches
 folium.CircleMarker([r.Latitude,r.Longitude],radius=11,color='#166d8c',fill=True,fill_opacity=.85,popup=f'{r.LaunchSite}: {r.successes}/{r.launches} ({pct:.1f}%)').add_to(m)
m.save('launch_sites_folium_from_notebook.html')
display(m)
```

```python
a=g[g.LaunchSite=='KSC LC 39A'].iloc[0];b=g[g.LaunchSite=='CCAFS SLC 40'].iloc[0]
def haversine(lat1,lon1,lat2,lon2):
 R=6371.0; p1,p2=map(math.radians,[lat1,lat2]);dp=math.radians(lat2-lat1);dl=math.radians(lon2-lon1)
 return 2*R*math.asin(math.sqrt(math.sin(dp/2)**2+math.cos(p1)*math.cos(p2)*math.sin(dl/2)**2))
print('KSC to CCAFS, straight line (km):',round(haversine(a.Latitude,a.Longitude,b.Latitude,b.Longitude),1))
folium.PolyLine([(a.Latitude,a.Longitude),(b.Latitude,b.Longitude)],color='orange',weight=3,tooltip='Great circle straight-line distance').add_to(m)
m.save('launch_sites_folium_with_distance.html')
```

Offline limitation: the map intentionally avoids external basemap tiles because tile access was blocked. A coastline-enhanced HTML map is provided separately in project supporting evidence.
