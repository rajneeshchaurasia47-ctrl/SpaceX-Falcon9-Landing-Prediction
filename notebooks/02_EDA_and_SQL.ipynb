# EDA and SQL analysis (90-row labelled dataset)
Data: IBM Skills Network `dataset_part_2.csv` (90 rows). Remote API/Wikipedia collection is not claimed successful.


```python
import pandas as pd, sqlite3, matplotlib.pyplot as plt
x=pd.read_csv('dataset_part_2.csv')
con=sqlite3.connect(':memory:')
x.to_sql('launches',con,index=False,if_exists='replace')
queries={'sites':"SELECT LaunchSite, COUNT(*) n, SUM(Class) successes, ROUND(100.0*AVG(Class),1) success_pct FROM launches GROUP BY LaunchSite ORDER BY n DESC",'orbits':"SELECT Orbit, COUNT(*) n, SUM(Class) successes, ROUND(100.0*AVG(Class),1) success_pct FROM launches GROUP BY Orbit ORDER BY n DESC",'reflight':"SELECT CASE WHEN Flights>1 THEN 'Reused' ELSE 'First flight' END flight_type, COUNT(*) n, ROUND(100.0*AVG(Class),1) success_pct FROM launches GROUP BY flight_type",'time':"SELECT SUBSTR(Date,1,4) year, COUNT(*) n, ROUND(100.0*AVG(Class),1) success_pct FROM launches GROUP BY year ORDER BY year"}
for name,query in queries.items():
 print('\n'+name+'\n'+query)
 print(pd.read_sql_query(query,con).to_string(index=False))
```

```python
(x.groupby('LaunchSite').Class.mean().mul(100).sort_values().plot.barh(figsize=(8,4),title='Landing success by launch site'))
plt.xlabel('Landing success (%)');plt.tight_layout();plt.show()
```

```python
(x.groupby('Orbit').Class.agg(['mean','size']).query('size >= 5')['mean'].mul(100).sort_values().plot.barh(figsize=(8,4),title='Landing success by orbit, n >= 5'))
plt.xlabel('Landing success (%)');plt.tight_layout();plt.show()
```

Dataset caution: this 90-row dataset is different from the IBM SQL lab 101-row `Spacex.csv`. Do not use `Class` as a replacement for that lab's detailed `Landing _Outcome` column.
