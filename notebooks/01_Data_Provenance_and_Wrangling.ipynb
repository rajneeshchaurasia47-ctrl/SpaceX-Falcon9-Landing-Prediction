## Collection sources and limitations

- Wikipedia frozen list (2021): https://en.wikipedia.org/w/index.php?title=List_of_Falcon_9_and_Falcon_Heavy_launches&oldid=1027686922
- Original SQL lab CSV: https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/labs/module_2/data/Spacex.csv
- API requests previously returned HTTP 522/525 and scraping HTTP 403. The supplied local CSV, not fetched API output, supports results below.


```python
from pathlib import Path
import pandas as pd
p=Path('dataset_part_2.csv')
df=pd.read_csv(p)
print('rows:',len(df),'columns:',len(df.columns))
print('Class counts:',df.Class.value_counts().sort_index().to_dict())
print('missing:',df.isna().sum()[lambda s:s.gt(0)].to_dict())
print('sites:',df.LaunchSite.unique().tolist())
```

```python
for field in ['LaunchSite','Orbit']:
    print('\n',field)
    print(df.groupby(field).Class.agg(launches='size',successes='sum',success_rate='mean').sort_values('launches',ascending=False).round(3).to_string())
```

Data preparation: preserve `Class` as the outcome; drop ID fields and `LandingPad` from ML features; impute and one-hot encode predictors using training-fold-only scikit-learn pipelines. Missing `LandingPad` count was 26 of 90 in the verified dataset.
